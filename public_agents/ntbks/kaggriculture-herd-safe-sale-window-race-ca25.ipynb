```python
from pathlib import Path
import base64
import hashlib
import io
import tarfile

ARCHIVE_BYTES = base64.b64decode("")
EXPECTED_ARCHIVE_SHA256 = "fea11a9e3e3d85298a29f0c5943a5c6e7bd9b64ed7d5d43ae49e4c74ff79075b"
EXPECTED_MAIN_SHA256 = "c4b72e64acede4053c4b41d797afb562ce2aeadf5d9d0cf58e18aed9baa33c9e"
EXPECTED_LICENSE_SHA256 = "cfc7749b96f63bd31c3c42b5c471bf756814053e847c10f3eb003417bc523d30"
EXPECTED_NOTICE_SHA256 = "82f3cbdc5ee3b4dbbad054b2ba703a8370b6431af573adf10ad92964421e6ea7"

def sha256(data):
    return hashlib.sha256(data).hexdigest()

assert sha256(ARCHIVE_BYTES) == EXPECTED_ARCHIVE_SHA256
with tarfile.open(fileobj=io.BytesIO(ARCHIVE_BYTES), mode="r:gz") as bundle:
    assert bundle.getnames() == ["LICENSE.txt", "NOTICE.txt", "main.py"]
    license_bytes = bundle.extractfile("LICENSE.txt").read()
    notice_bytes = bundle.extractfile("NOTICE.txt").read()
    main_bytes = bundle.extractfile("main.py").read()
assert sha256(license_bytes) == EXPECTED_LICENSE_SHA256
assert sha256(notice_bytes) == EXPECTED_NOTICE_SHA256
assert sha256(main_bytes) == EXPECTED_MAIN_SHA256
compile(main_bytes, "main.py", "exec")
Path("submission.tar.gz").write_bytes(ARCHIVE_BYTES)
print("PASS: wrote submission.tar.gz", len(ARCHIVE_BYTES), "bytes")
print("archive SHA-256:", EXPECTED_ARCHIVE_SHA256)
print("main.py SHA-256:", EXPECTED_MAIN_SHA256)
```
